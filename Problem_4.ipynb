```python
raw_df=df.copy()
clean_df=df.copy()

original_columns=clean_df.columns.tolist()
print(original_columns)
print("Original shape:",clean_df.shape)
```

```python
print("Gender before Cleaning:")
print(clean_df["Gender"].value_counts(dropna=False)) # Gender and Hired are categorical data
```

```python
print("Hired before Cleaning:")
print(clean_df["Hired"].value_counts(dropna=False)) #value_counts counts the unique entries from the given data set.
```

```python
clean_df["Gender"]=clean_df["Gender"].replace({"M":"Male","male":"Male"})
```

```python
clean_df["Hired"]=clean_df["Hired"].replace({"Y":"Yes","YES":"Yes"})
```

```python

print("Gender after Cleaning")
print(clean_df["Gender"].value_counts(dropna=False))
```

```python
print("Hired after Cleaning:")
print(clean_df["Hired"].value_counts(dropna=False))
```

```python
percentage_columns = [
    "PU_Marks_Pct",
    "Attendance_Pct",
    "Engineering_Marks_Pct",
    "Aptitude_Test_Pct",
]

percentage_columns
```

```python
invalid_counts = {} #Empty Dictionary

for column in percentage_columns:
    invalid_mask = ~clean_df[column].between(0, 100) & clean_df[column].notna()
    invalid_counts[column] = int(invalid_mask.sum()) #.sum() gives the number of true values from the column.

pd.Series(invalid_counts, name="Invalid Values")
```

```python
for column in percentage_columns:
    invalid_mask = ~clean_df[column].between(0, 100) & clean_df[column].notna()
    clean_df.loc[invalid_mask,column] =np.nan
    # print(clean_df)

clean_df[percentage_columns].isna().sum().rename("Marks After Invalid Values were Marked")
```

```python
percentage_medians = clean_df[percentage_columns].median()
clean_df[percentage_columns] = clean_df[percentage_columns].fillna( percentage_medians)

percentage_medians.round(2).rename("Median Used")
```

```python
count_columns = ["Internships_Completed", "Hackathons_Participated"]
print(clean_df[count_columns].count())


for column in count_columns:
    median_value = clean_df[column].median()
    clean_df[column] = (clean_df[column].fillna(median_value).round().astype(int))
    print(f"{column}: median used = {median_value:.0f}")
    print(type(median_value))
    
```
